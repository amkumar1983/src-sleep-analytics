```python

## Read details send by ADF
pipeline_name = dbutils.widgets.get("pipeline_name")
run_id = dbutils.widgets.get("run_id")
activity_name = dbutils.widgets.get("activity_name")
status = dbutils.widgets.get("status")
trigger_name = dbutils.widgets.get("trigger_name")
execution_time = dbutils.widgets.get("execution_time")
error_code = dbutils.widgets.get("error_code")
error_message = dbutils.widgets.get("error_message")
source_file = dbutils.widgets.get("source_file")
```

```python
## create log content
from datetime import datetime

log = {
    "pipeline_name": "pipeline_name",
    "run_id": "run_id",
    "activity_name": "activity_name",
    "status": "status",
    "trigger_name": "trigger_name",
    "execution_time": "execution_time",
    "error_code": "error_code",
    "error_message": "error_message",
    "source_file": "source_file"
}
```

```python
import json
from datetime import datetime

json_log = json.dumps(log, indent=4)

today = datetime.utcnow()

folder = f"abfss://audit@stsleepanalytics.dfs.core.windows.net/pipeline_logs/failure/{today.year}/{today.month:02}/{today.day:02}"

file_name = f"{run_id}.json"

dbutils.fs.put(
    f"{folder}/{file_name}",
    json_log,
    True
)
```
